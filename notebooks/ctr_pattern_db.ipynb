# CTR 패턴 분석 (DB / Parquet)

**데이터**: 광고 클릭 로그 (Parquet 또는 PostgreSQL / MongoDB)
**처리**: Python + pandas
**출력**: Streamlit 대시보드 (`app_streamlit.py`)에서 시각화

## 데이터 소스
- `parquet`: `data/train_sample.parquet` (기본)
- `postgres`: PostgreSQL `ad_click_logs` 테이블
- `mongodb`: MongoDB `ctr_analysis.ad_click_logs` 컬렉션

```python
import sys
sys.path.insert(0, '..')

from src.data_loader import load_click_logs
from src.ctr_analysis import compute_ctr_patterns, get_overall_stats, get_time_patterns
import pandas as pd

# 데이터 소스: parquet | postgres | mongodb
SOURCE = "parquet"
df = load_click_logs(source=SOURCE, path="../data/train_sample.parquet", n_rows=100_000)
print(f"로드 완료: {df.shape}")
```

```python
# 전체 통계
stats = get_overall_stats(df)
for k, v in stats.items():
    print(f"{k}: {v}")
```

```python
# 차원별 CTR 패턴
patterns = compute_ctr_patterns(df)
for dim, ctr_df in patterns.items():
    print(f"\n=== {dim} ===")
    display(ctr_df.head(10))  # Jupyter display; repl에서 실행 시 print(ctr_df) 사용
```

```python
# 시간대·요일별 패턴
hour_ctr, day_ctr = get_time_patterns(df)
if not hour_ctr.empty:
    display(hour_ctr)
if not day_ctr.empty:
    display(day_ctr)
```
